```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# study Hours 
X = [[1], [2], [3], [4], [5], [6], [7] , [8]]

# 0 = Fail, 1 = Pass 
Y = [0,0,0,0,1,1,1,1]

# Create model 
model = LogisticRegression()

# Trian the model 
model.fit(X,Y)

# prediction 
print("2 hours: ", model.predict([[2]]))

print("5 hours: ", model.predict([[5]]))

# plot 
plt.scatter(X,Y, color="yellow")

# Decision boundary 
boundary = -model.intercept_[0] / model.coef_[0][0]
plt.axvline(boundary, color="blue")

plt.xlabel("Study Hours")
plt.ylabel("Result")
plt.title("Logistic Regression")

plt.show()
```
