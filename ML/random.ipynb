```python
# 1. Libraries import kara
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 2. Sample Data
# Hours_Study, Attendance, Previous_Marks -> Pass/Fail
data = {
    'Hours_Study': [2, 5, 1, 8, 4, 9, 3, 7, 6, 10],
    'Attendance': [40, 60, 30, 90, 70, 95, 50, 85, 75, 100],
    'Previous_Marks': [35, 55, 25, 80, 60, 88, 45, 78, 65, 92],
    'Result': ['Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass']
}

df = pd.DataFrame(data)
print("Data:\n", df)

# 3. spliting data into X and y
X = df[['Hours_Study', 'Attendance', 'Previous_Marks']]  # Features
print("\t")
y = df['Result']  # Target

# 4. Train ani Test data madhe split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 5. Create Random Forest Model
rf_model = RandomForestClassifier(
    n_estimators=100,   # 100 tree
    random_state=42,    # Result is same
    max_depth=5         # tree depth
)

# 6. Train the Model
rf_model.fit(X_train, y_train)

# 7. Predict the Model
y_pred = rf_model.predict(X_test)

# 8. Accuracy
print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nReport:\n", classification_report(y_test, y_pred))
```
