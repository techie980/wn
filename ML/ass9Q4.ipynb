```python
import time 
from datetime import datetime 

# Users and their permission 
users = {
    "Admin": "Full Access",
    "Manager": "Edit",
    "Team Member": "View Only"
}

# Display Permissions 
print("===== USER PERMISSIONS =====")
for user, permission in users.items():
    print(user, ":", permission)

# Data Refresh Function 
def refresh_data():
    print("\nRefresh data...")
    time.sleep(2)
    print("Data Refreshed Successfully")
    print("Refresh Time:", datetime.now())

# Refresh Data 
refresh_data()

# Check permission 
user = "Team Member"
print("\n===== Access Check =====")
print("User:", user)
print("Permission:", users[user])

if (users[user] == "Full Access"):
    print("User can view, edit and manage data")
elif (users[user] == "Edit"):
    print("User can view and edit data")
else:
    print("User can only view data")
    
```
