```python
# Create a 1-D array 
import numpy as np 
arr1 = np.array([10, 20, 30])
arr2 = np.array([
    [40, 50, 60],
    [70, 80, 90]
])

# Convert 1-D array into a 2-D row 
arr1 = arr1.reshape(1, 3)

# Continue arrays row-wise 
result = np.concatenate((arr1, arr2), axis=0)
print("Combined Array:")
print(result)
```
