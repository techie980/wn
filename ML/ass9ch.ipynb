```python
# Bar chart1
import matplotlib.pyplot as plt 
students = ["Ram", "Bhim", "Ajay", "Amit", "Diya"]
marks = [60,65,70,90,95]
plt.bar(students, marks)
plt.title("Students Marks")
plt.xlabel("Student Names")
plt.ylabel("Student Marks")

plt.show()
```

```python
# Bar Chart2
import matplotlib.pyplot as plt 
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 120, 180, 200]
plt.bar(months, sales, color="purple")
plt.title("Month Sales")
plt.xlabel("Months")
plt.ylabel("Sales IN units")

plt.show()
```

```python
# Line Chart
import matplotlib.pyplot as plt 
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [80, 100, 50, 90, 150]
plt.plot(months, sales, marker="o", color="red")
plt.title("Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales IN units")

plt.show()
```

```python
# Pie chart 
import matplotlib.pyplot as plt 
subjects = ["Maths", "Science", "English", "Hindi", "History"]
marks = [35, 60, 65, 80, 45]
plt.pie(marks, labels=subjects, autopct="%1.1f%%")
plt.title("Marks of Subjects")


plt.show()

```

```python
# Area Chart
import matplotlib.pyplot as plt 
momths = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [80, 100, 50, 90, 150]
plt.plot(months, sales)
plt.fill_between(months, sales, alpha=0.5)
plt.title("Sales Area chart")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# Scatter Chart
import matplotlib.pyplot as plt 
hours = [1,2,3,4,5,6]
marks = [55, 60, 45, 70, 87, 35]
plt.scatter(hours, marks, color="purple")
plt.title("Study Hours vs Marks")
plt.xlabel("Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# Histogram Chart 
import matplotlib.pyplot as plt 
marks = [45, 55, 60, 35, 23, 78, 90, 56, 54, 37, 89, 76]
plt.hist(marks, bins=5, color="blue")
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
# seaborn 
import seaborn as sns 
import matplotlib.pyplot as plt 
students = ["A", "B", "C", "D", "E"]
marks = [50, 65, 70, 80, 90]
sns.barplot(x=students, y=marks)
plt.title("Student Marks")
plt.show()
```
