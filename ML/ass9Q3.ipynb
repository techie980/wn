```python
import pandas as pd 
import matplotlib.pyplot as plt 

data = {
    "Student": ["Aniket", "Raj", "Shubham", "Dinesh", "Vinit"],
    "Marks": [65, 82, 56, 91, 67]
}

# Create table 
df = pd.DataFrame(data)

# Display report 
print("===================================================")
print("                    Student REPORT                 ")
print("===================================================")
print(df)

# Calculate average
average = df["Marks"].mean()
print("\nAverage Marks:", average)

# Find Highest Marks 
top_student = df.loc[df["Marks"].idxmax()]

print("Top Student:", top_student["Student"])
print("Highest Marks:", top_student["Marks"])

# Create dashboard chart 
plt.figure(figsize=(7, 4))
plt.bar(df["Student"], df["Marks"], color="skyblue")

plt.title("Student Performance Dashboard")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(0, 100)

plt.show()

# Save report 
df.to_csv("Students_report.csv", index=False)
print("\nReport saved successfully")
print("The report can be shared with team members.")
```

```python
# Sharing Reports and dashboards with team members 
import pandas as pd 
import matplotlib.pyplot as plt 

# Sample ML results 
data = {
    "Model": ["Decision Tree", "Random Forest", "Logistic Regression"],
    "Accuracy": [85, 92, 88]
}

# Create a DataFrame 
df = pd.DataFrame(data)

# Display Dashboard 
print("===================================================")
print("                ML REPORT DASHBOARD                ")
print("===================================================")
print(df)

print("\nBest Model:")
base_model = df.loc[df["Accuracy"].idxmax()]

print("Model:", base_model["Model"])
print("Accuracy:", str(base_model["Accuracy"]) + "%")

# Create a chart 
plt.figure(figsize=(7, 4))
plt.bar(df["Model"], df["Accuracy"], color=["blue", "green", "orange"])

plt.title("Model Accuracy")
plt.xlabel("Machine Learning Model")
plt.ylabel("Accuracy (%)")
plt.ylim(0, 100)

plt.show()

# Save report to a csv file 
df.to_csv("ML_Report.csv", index="False")
print("\nReport saved")
```
