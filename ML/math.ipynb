```python
# i) Create a 10x10 array with random values 
import numpy as np 
arr = np.random.rand(10, 10)
print("10x10 Random Array")
print(arr)
print("\nMinimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Mean:", np.mean(arr))
print("Standard Deviation:", np.std(arr))

# ii) Dot Product (Matrix multiplication)
A = np.random.rand(5, 3)
B = np.random.rand(3, 2)

C = np.dot(A, B)

print("\n5x3 Matrix A:")
print(A)

print("\n3x2 Matrix B:")
print(B)

print("\nDot Product (A x B):")
print(C)
```
