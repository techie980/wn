```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.cluster import KMeans, DBSCAN

#1. Create concentric cricles dataset
X,_ = make_circles(n_samples=400, factor=0.5, noise=0.05, random_state=42)


#2. Fit K-Means VS DBSCAN 
kmeans_labels = KMeans(n_clusters=2, random_state=42, n_init="auto").fit_predict(X)
dbscan_labels = DBSCAN(eps=0.15, min_samples=5).fit_predict(X)

#3. Plot side-by-side
fig,axes = plt.subplots(1,2, figsize=(10,4)) 

# K-Means Plot 
axes[0].scatter(X[:,0], X[:,1], c=kmeans_labels, cmap="coolwarm", edgecolors="k")

axes[0].set_title("K-Means (Fails on shapes)")

# DBSCAN plot 
axes[1].scatter(X[:,0], X[:,1], c=dbscan_labels, cmap="coolwarm", edgecolors="k")

axes[1].set_title("DBSCAN (Succeeds on density)")

plt.show()
```
