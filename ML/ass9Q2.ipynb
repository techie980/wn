```python
# Simple Bar Chart
import matplotlib.pyplot as plt 
X = ["A", "B", "C", "D"]
Y = [10, 20, 15, 25]
plt.bar(X, Y)
plt.show()
```

```python
# Formatting the chart 
import matplotlib.pyplot as plt 
X = ["A", "B", "C", "D"]
Y = [10, 20, 15, 25]
plt.bar(X, Y, color="green")
plt.title("Simple Bar Chart")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
# Customizing the chart 
import matplotlib.pyplot as plt 
X = ["A", "B", "C"]
Y = [20, 40, 30]
plt.bar(X, Y, color=["orange", "blue", "red"])
plt.title("Student Marks")
plt.xlabel("Names")
plt.ylabel("Marks")
plt.show()
```

```python
# Save and publish 
import matplotlib.pyplot as plt 
X = ["A", "B", "C"]
Y = [20, 40, 30]
plt.bar(X, Y)
plt.title("Student Marks")
plt.savefig("report.png")
plt.show()
```
