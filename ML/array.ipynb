```python
# Create a 4X2 integer Array and print shape Dimension, and Data Type
import numpy as np 

# Create a 4x2 integer array 
arr = np.array([
    [10,20],
    [30,40],
    [50,60],
    [70,80]
])

# Print array 
print("Array:")
print(arr)

# Print Shape 
print("Shape:", arr.shape)

# Print Dimension 
print("Dimension:", arr.ndim)

# Print data type 
print("Data Type:", arr.dtype) 

```
