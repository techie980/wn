```python
import numpy as np 
# Create a 4x3 Numpy Array 
arr = np.arange(1, 13).reshape(4,3)
print("Original Array")
print(arr)

# i) Extract the second row 
second_row = arr[:2, :2]
print("\nSecond Row")
print(second_row)

# ii) Extract a 2x2 matrix from the top-left corner 
sub_matrix = arr[:2, :2]
print("\nTop-left 2x2 SubMatrix")
print(sub_matrix)

# iii) Extract the last two rows 
last_rows = arr[2:, :]
print("\nLast Two Rows:")
print(last_rows)

# iv) Extract the second column 
second_column = arr[:, 1]
print("\nSecond Column")
print(second_column)

# v) Extract alternate elements 
alternate = arr[::2, :]
print("\nAlternate Rows:")
print(alternate)

```
